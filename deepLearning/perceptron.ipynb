```python
import torch
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt 
import numpy as np
```

```python
np.random.seed(46)
X, Y = make_classification(n_features=2, n_redundant=0, n_informative=1, n_clusters_per_class=1)
```

```python
# Criar função que recebe os ceficientes e plota uma reta y = ax + b

def calcPoint(x,y):
    plt.plot(x, y)

def linearPlot(w1, w2, b):
    
    # y = (ax + c)/b 
    # w1x + w2y + b = 0

    x = np.linspace(-4, 4, 50) # define 50 points between -4 and 4, the x space
    y = (-w1*x - b)/w2 # define y 
    plt.xlim(-4, 4)
    plt.ylim(-4,4)
    plt.scatter(X[:,0], X[:,1], marker='o', c=Y)
    plt.axvline(0, 0, 1, color="black", linewidth=0.5)
    plt.axhline(0, 0, 1, color="black", linewidth=0.5)
    plt.plot(x, y)


w1 = 8
w2 = 2
b = -1

linearPlot(w1, w2, b)


```

```python
def classify(px, py):
    value = w1*px + w2*py + b

    if(value <= 0):
        return 0, "purple"
    else:
        return 1, "yellow"

p1 = [-3,3]
classification, color = classify(p1[0], p1[1])
print(classification, color)

linearPlot(w1, w2, b)
plt.plot(p1[0], p1[1], marker="^", color=color, markersize=10) 
```

```python
def check_prediction(p, val):
    prediction = classify(p[0], p[1])[0]
    if(prediction == val):
        return 1
    return 0

score = 0
for i in range(len(X)):
    score+=check_prediction(X[i], Y[i])
    
precision = score/len(X)
print(f'Score for linear classifier: {precision}')
```
